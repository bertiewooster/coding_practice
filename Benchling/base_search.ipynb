## Sets

```python
s:set = set()
s.add("A")
s.add("B")
```

```python
s2:set = set()
s2.add("B")
s2.add("C")
```

```python
s.intersection(s2)
```

```python
s - s2
```

```python
s.union(s2)
```

## Combinations

```python
from itertools import product
```

```python
p = product([1, 2], [3, 4])
```

```python
list(p)
```

```python
items = [["A", "G"], ["C", "T"], ["C"]]
```

```python
list(product(*items))
```
